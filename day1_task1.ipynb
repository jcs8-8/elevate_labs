```python
# ================================================================
# TASK 1: DATA CLEANING & PREPROCESSING
# TOPIC: AGENTIC AI AGENT PERFORMANCE DATASET
# ================================================================

# ------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler

print("=" * 60)
print("AGENTIC AI DATA CLEANING & PREPROCESSING")
print("=" * 60)

# ------------------------------------------------
# 2. CREATE AGENTIC AI DATASET
# ------------------------------------------------

np.random.seed(42)

n = 1000

agent_types = [
    "Research Agent",
    "Coding Agent",
    "Customer Support Agent",
    "Data Analysis Agent",
    "Planning Agent"
]

task_types = [
    "Research",
    "Coding",
    "Customer Support",
    "Data Analysis",
    "Planning"
]

complexity_levels = [
    "Low",
    "Medium",
    "High"
]

data = {
    "agent_id": range(1, n + 1),

    "agent_type": np.random.choice(
        agent_types, n
    ),

    "task_type": np.random.choice(
        task_types, n
    ),

    "task_complexity": np.random.choice(
        complexity_levels,
        n,
        p=[0.3, 0.5, 0.2]
    ),

    "execution_time": np.random.normal(
        45, 15, n
    ),

    "tool_calls": np.random.poisson(
        8, n
    ),

    "token_usage": np.random.normal(
        2500, 800, n
    ),

    "task_quality_score": np.random.normal(
        80, 10, n
    ),

    "human_intervention": np.random.choice(
        ["Yes", "No"],
        n,
        p=[0.25, 0.75]
    ),

    "success": np.random.choice(
        ["Success", "Failure"],
        n,
        p=[0.8, 0.2]
    ),

    "cost": np.random.normal(
        0.15, 0.05, n
    )
}

df = pd.DataFrame(data)

print("\n1. Dataset created successfully!")
print("Dataset shape:", df.shape)

# ------------------------------------------------
# 3. DISPLAY ORIGINAL DATA
# ------------------------------------------------

print("\n2. First 5 rows of the dataset:")
display(df.head())

# ------------------------------------------------
# 4. DATASET INFORMATION
# ------------------------------------------------

print("\n3. Dataset Information:")
df.info()

# ------------------------------------------------
# 5. STATISTICAL SUMMARY
# ------------------------------------------------

print("\n4. Statistical Summary:")
display(df.describe())

# ------------------------------------------------
# 6. INTRODUCE MISSING VALUES
# ------------------------------------------------

missing_columns = [
    "execution_time",
    "tool_calls",
    "token_usage",
    "task_quality_score",
    "task_type"
]

for column in missing_columns:

    missing_indices = np.random.choice(
        df.index,
        size=20,
        replace=False
    )

    df.loc[
        missing_indices,
        column
    ] = np.nan

print("\n5. Missing values introduced for demonstration.")

# ------------------------------------------------
# 7. CHECK MISSING VALUES
# ------------------------------------------------

print("\n6. Missing values before cleaning:")
print(df.isnull().sum())

# ------------------------------------------------
# 8. VISUALIZE MISSING VALUES
# ------------------------------------------------

plt.figure(figsize=(12, 5))

sns.heatmap(
    df.isnull(),
    cbar=False
)

plt.title("Missing Values Before Cleaning")
plt.show()

# ------------------------------------------------
# 9. HANDLE NUMERICAL MISSING VALUES
# ------------------------------------------------

numerical_columns = [
    "execution_time",
    "tool_calls",
    "token_usage",
    "task_quality_score",
    "cost"
]

for column in numerical_columns:

    df[column] = df[column].fillna(
        df[column].median()
    )

print("\n7. Numerical missing values handled using median.")

# ------------------------------------------------
# 10. HANDLE CATEGORICAL MISSING VALUES
# ------------------------------------------------

df["task_type"] = df["task_type"].fillna(
    df["task_type"].mode()[0]
)

print("Categorical missing values handled using mode.")

# ------------------------------------------------
# 11. VERIFY MISSING VALUES
# ------------------------------------------------

print("\n8. Missing values after cleaning:")
print(df.isnull().sum())

# ------------------------------------------------
# 12. CHECK CATEGORICAL FEATURES
# ------------------------------------------------

print("\n9. Categorical columns:")

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

print(categorical_columns.tolist())

# ------------------------------------------------
# 13. ENCODE CATEGORICAL VARIABLES
# ------------------------------------------------

df = pd.get_dummies(
    df,
    columns=[
        "agent_type",
        "task_type",
        "task_complexity",
        "human_intervention",
        "success"
    ],
    drop_first=True
)

print("\n10. Categorical variables encoded successfully.")

# ------------------------------------------------
# 14. CONVERT BOOLEAN TO INTEGER
# ------------------------------------------------

boolean_columns = df.select_dtypes(
    include=["bool"]
).columns

df[boolean_columns] = df[
    boolean_columns
].astype(int)

print("Boolean values converted to 0 and 1.")

# ------------------------------------------------
# 15. DISPLAY DATA AFTER ENCODING
# ------------------------------------------------

print("\n11. Dataset after categorical encoding:")
display(df.head())

# ------------------------------------------------
# 16. OUTLIER VISUALIZATION - EXECUTION TIME
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["execution_time"]
)

plt.title(
    "Execution Time - Before Outlier Removal"
)

plt.xlabel("Execution Time")

plt.show()

# ------------------------------------------------
# 17. OUTLIER VISUALIZATION - TOKEN USAGE
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["token_usage"]
)

plt.title(
    "Token Usage - Outlier Detection"
)

plt.xlabel("Token Usage")

plt.show()

# ------------------------------------------------
# 18. FIND OUTLIERS USING IQR
# ------------------------------------------------

def find_outliers(data, column):

    Q1 = data[column].quantile(0.25)

    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR

    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    return (
        outliers,
        lower_limit,
        upper_limit
    )


execution_outliers, lower_execution, upper_execution = \
    find_outliers(
        df,
        "execution_time"
    )

print("\n12. Outlier Detection:")
print(
    "Number of execution time outliers:",
    len(execution_outliers)
)

print(
    "Lower limit:",
    round(lower_execution, 2)
)

print(
    "Upper limit:",
    round(upper_execution, 2)
)

# ------------------------------------------------
# 19. REMOVE OUTLIERS
# ------------------------------------------------

df = df[
    (df["execution_time"] >= lower_execution) &
    (df["execution_time"] <= upper_execution)
]

print("\n13. Outliers removed successfully.")

print(
    "Dataset shape after outlier removal:",
    df.shape
)

# ------------------------------------------------
# 20. BOX PLOT AFTER OUTLIER REMOVAL
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["execution_time"]
)

plt.title(
    "Execution Time - After Outlier Removal"
)

plt.xlabel("Execution Time")

plt.show()

# ------------------------------------------------
# 21. STANDARDIZATION
# ------------------------------------------------

numerical_features = [
    "execution_time",
    "tool_calls",
    "token_usage",
    "task_quality_score",
    "cost"
]

scaler = StandardScaler()

df[numerical_features] = scaler.fit_transform(
    df[numerical_features]
)

print("\n14. Numerical features standardized successfully.")

# ------------------------------------------------
# 22. VERIFY STANDARDIZATION
# ------------------------------------------------

print("\n15. Mean after standardization:")

print(
    df[numerical_features].mean().round(2)
)

print("\nStandard deviation after standardization:")

print(
    df[numerical_features].std().round(2)
)

# ------------------------------------------------
# 23. CORRELATION HEATMAP
# ------------------------------------------------

plt.figure(figsize=(14, 10))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f"
)

plt.title(
    "Correlation Heatmap - Agentic AI Dataset"
)

plt.show()

# ------------------------------------------------
# 24. FINAL DATASET
# ------------------------------------------------

print("\n" + "=" * 60)
print("FINAL CLEANED DATASET")
print("=" * 60)

print("\nFinal dataset shape:")
print(df.shape)

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nFinal dataset:")
display(df.head(10))

# ------------------------------------------------
# 25. SAVE AND DOWNLOAD CLEANED DATASET
# ------------------------------------------------

file_name = "agentic_ai_cleaned.csv"

df.to_csv(
    file_name,
    index=False
)

print("\n" + "=" * 60)
print("PREPROCESSING COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("\nCleaned dataset saved as:")
print(file_name)

# Download file in Google Colab
from google.colab import files

files.download(file_name)
```
